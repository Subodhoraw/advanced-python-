# What is PySpark?

**PySpark** is the Python API for **Apache Spark**, a distributed computing framework. It lets you process **very large datasets** (millions or billions of rows) across multiple machines.

Instead of using Pandas:

```python
import pandas as pd

df = pd.read_csv("employees.csv")
```

you use PySpark:

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName("Employees").getOrCreate()

df = spark.read.csv("employees.csv", header=True)
```

# Regex in PySpark

These are used a lot in ETL and data cleaning.

Extract email:

```python
from pyspark.sql.functions import regexp_extract

df.withColumn(
    "email",
    regexp_extract("text", r'[\w.-]+@[\w.-]+', 0)
)
```

---

Replace phone number:

```python
from pyspark.sql.functions import regexp_replace

df.withColumn(
    "text",
    regexp_replace("text", r"\d{10}", "<PHONE>")
)
```
