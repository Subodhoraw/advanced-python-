### You are developing a statistical application that requires calculating the average of a list of numerical inputs. Users will provide the input, which may sometimes include invalid data types. Your task is to create a robust function capable of handling these scenarios and returning a valid average or appropriate error messages.


Problem Description
Implement the function calculate_average() that

Accepts a single argument numbers, a list containing numerical values

Returns a float representing the average of the numbers in the list.

If the list contains any non-numeric values, the exception handling logic for TypeError must print the message Error: Invalid type found in the list.

If the input list is empty, the exception handling logic for ZeroDivisionError must print the message Error: Invalid type found in the list.



```python
def calculate_average(numbers):
    try:
        average = sum(numbers) / len(numbers)
        return average

    except TypeError:
        print("Error: Invalid type found in the list.")

    except ZeroDivisionError:
        print("Error: Invalid type found in the list.")

numbers = [10, 20, 'thirty', 40]
print(calculate_average(numbers))



```

In various applications, mathematical operations are essential. Division, in particular, is a common operation that can lead to errors if not handled properly, such as division by zero or the use of non-numeric values. A reliable function for division is crucial for software applications that perform financial calculations, scientific computations, or simple arithmetic operations.


Problem Description
Implement the function safe_divide() that

Accepts two arguments
numerator, an int/float value that is the numerator for the division operation, and

denominator, an int/float value that is the denominator for the division operation

Returns a float value that is the quotient of the division.

Notes
Ensure that both inputs are numeric (either integers or floats). If the inputs are invalid (non-numeric), returns None and prints an error message.

Handle division by zero gracefully. If a division by zero occurs, returns None and prints an error message.

Any unexpected errors should also be captured and printed. In case of any unexpected errors, returns None and prints an error message.

```python
def safe_divide(numerator,denominator):
    try:
        divisable = numerator / denominator
        return divisable
    except ZeroDivisionError :
        print(f" error message")
        return None
    except TypeError:
        print("Both input should be numeric ")
        return None
    except Exception as e:
        print(f" errors on it",e)
        return None
     



```
