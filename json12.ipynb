## In the modern digital world, many businesses need to keep track of customer orders in a structured format. Converting order details into a standardised format, such as JSON, allows for easy data transfer and storage. This is particularly useful in food delivery services, where customers place orders online that need to be processed efficiently.


Problem Description
Implement the function order_to_json() that

Accepts three arguments
customer_name, a string that represents the name of the customer, and

items, a list of strings representing a list of ordered food items, and

total_price, a float value representing the total price of the order

Returns a JSON string representing the order details as seen in the sample outputs below.

Notes
The function should handle the order details even if the list of items is empty, returning a valid JSON string.


```python
def order_to_json(customer_name, items, total_price):
    
    order = {
        "customer_name": customer_name,
        "items": items,
        "total_price": total_price
    }

    return json.dumps(order)
   
customer_name = "John Doe"

items = ["Burger", "Fries", "Soda"]

total_price = 12.99

print(order_to_json(customer_name, items, total_price)) 
```

### Use Case
 In a survey analysis application, it's crucial to filter out responses based on customer satisfaction scores. This allows businesses to identify which respondents are satisfied and which are not, making it easier to analyze feedback and make necessary improvements.


Problem Description
Implement the function filter_responses_by_score() that

Accepts two arguments
survey_json, a JSON-formatted string containing survey data, and

min_score, a number denotes the minimum satisfaction score threshold

Returns a JSON-formatted string with only the responses that meet or exceed the minimum score

Notes
If no responses meet the minimum score, return an empty list for responses.

```python
import json

def filter_responses_by_score(survey_json, min_score):

    survey_dict = json.loads(survey_json)

    filtered = []

    for response in survey_dict["responses"]:

        if response["satisfaction_score"] >= min_score:
            filtered.append(response)

    survey_dict["responses"] = filtered

    return json.dumps(survey_dict)


survey_json = '''
{
    "survey_id": 101,
    "survey_name": "Customer Satisfaction Survey",
    "responses": [
        {"respondent_id": 1, "satisfaction_score": 4},
        {"respondent_id": 2, "satisfaction_score": 5},
        {"respondent_id": 3, "satisfaction_score": 3}
    ]
}
'''

print(filter_responses_by_score(survey_json, 4))
```
