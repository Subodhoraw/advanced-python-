```python
import json
{


    "survey_id": 101,


    "survey_name": "Customer Satisfaction Survey",


    "responses": [


        {"respondent_id": 1, "satisfaction_score": 4},


        {"respondent_id": 2, "satisfaction_score": 5},


        {"respondent_id": 3, "satisfaction_score": 3}


    ]


}


def calculate_average_satisfaction(survey_json_str):

    responses = json.load(survey_json_str) # loading json to turn this into dict 

    total = 0
    

    for response in responses:
        total += response["score"]

    return int(total / len(responses))

```
