````python
def find_depth(n):
    if not isinstance(n, dict) or not n:
        return 0
    return 1 + max(find_depth(value) for value in n.values())
````

````python
def find_depth(d):
    if not isinstance(d, dict) or not d:
        return 0

    return 1 + max(find_depth(value) for value in d.values())
````

Use Case
In a company's digital filing system, various departments need to manage their documents efficiently. The system uses nested dictionaries to store information about different departments and the documents they handle. Each department has specific categories, each containing a list of files. As the company evolves, the structure needs to accommodate changes in document management by allowing for updates and additions.


Problem Description
Perform updates directly on the nested dictionary using standard dictionary and list operations. The updates involve:

Update an Existing File:
Locate the department "Finance".

Navigate to the category "Taxes".

Replace "2021_Tax_Return.pdf" with "2022_Tax_Return.pdf".

Add a New Category:
Locate the department "Finance".

Add a new category "Budgets" with files ["Q1_Budget.pdf", "Q2_Budget.pdf"].

Add a New File:
Locate the department "Finance".

Add the file "Invoice_003.pdf" to the existing category "Invoices".



````python

filing_cabinet = {

    "Finance": {

        "Taxes": ["2021_Tax_Return.pdf", "2020_Tax_Return.pdf"],

        "Invoices": ["Invoice_001.pdf", "Invoice_002.pdf"]

    },

    "HR": {

        "Policies": ["Employee_Handbook.pdf", "Code_of_Conduct.pdf"],

        "Employee Records": ["Jon_Doe_Record.pdf", "Jane_Smith_Record.pdf"]

    }

}

filing_cabinet["Finance"]["Taxes"][0] = "2022_Tax_return.pdf" # update an existing file 
print(filing_cabinet)

filing_cabinet["Finance"]["Budget"] = ["Q1_Budget.pdf", 
    "Q2_Budget.pdf"]  # adding new category 
print(filing_cabinet)

filing_cabinet["Finance"]["Invoices"].append("Invoice_003.pdf")
print(filing_cabinet)

# adding new value 
filing_cabinet["Finance"]["Invoices"].append("Invoice_003.pdf")
print(filing_cabinet)

````
