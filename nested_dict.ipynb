employees
│

Use Case
Imagine you are working for a logistics company that needs to inform customers about the estimated time of arrival (ETA) for their parcels. Customers often want to know when they can expect their deliveries based on the days and additional hours specified by the company.


Problem Description
Implement the function calculate_eta() that

Accepts two arguments
delivery_days, an integer representing the number of days for delivery, and

delivery_hours, an integer representing additional hours for delivery

Returns the ETA as a string formatted in "YYYY-MM-DD HH:MM AM/PM"

````python
from datetime import datetime, timedelta
def calculate_eta(delivery_days,delivery_hours):
    current_time = datetime.now()
    eta = current_time + timedelta(
        days = delivery_days,hours= delivery_hours
    )
    return eta.strftime("%Y-%m-%d %I:%M %p")

````

Problem Description
Implement a function extract_date_components() that:

Accepts a string representing a date in the "YYYY-MM-DD" format.

Extracts the day of the week, month, and year from the provided date.

Returns a dictionary containing the keys day_of_week, month, and year, with corresponding values.


Notes
The day of the week should be returned as a full name (e.g., 'Monday', 'Tuesday').

The month should also be returned as a full name (e.g., 'January', 'February').

Ensure your function can handle any valid date string within the given format.



````python
from datetime import datetime

def extract_date_components(date_string):

    dt = datetime.strptime(date_string, "%Y-%m-%d")

    return {
        "day_of_week": dt.strftime("%A"),
        "month": dt.strftime("%B"),
        "year": dt.year
    }
````

Accepts a single argument radius, a float value that represents the radius of the circle.

Returns a tuple of two float values, first representing the circumference and the second representing the area of the circle.


Notes
The input radius will always be a positive float.

Be mindful of the precision of π; Python’s math library offers a good representation.

````python
import math

def circle_properties(radius):

    circumference = 2 * math.pi * radius
    area = math.pi * radius ** 2

    return (circumference, area)
````

Use Case
On the special day of their birthday, people often wish to know if today is their birthday. This feature is common in birthday reminder applications, personal assistant tools, and social networking sites where users can receive notifications on their special days.


Problem Description
Implement the function birthday_checker() that

Accepts two arguments
dob_str, a string representing date of birth of the user in the format "dd-mm-yyyy", and

current_date_str, a string representing current date in the format "dd-mm-yyyy"

Returns a string "Happy Birthday!" if it is the user's birthday, or "Your birthday is not today." otherwise

````python
from datetime import datetime

def birthday_checker(dob_str, current_date_str):

    dob = datetime.strptime(dob_str, "%d-%m-%Y")
    current = datetime.strptime(current_date_str, "%d-%m-%Y")

    if dob.day == current.day and dob.month == current.month:
        return "Happy Birthday!"
    else:
        return "Your birthday is not today."
````

Use Case
In many applications, it is essential to process date strings, extract their components, and provide them in a structured format. For example, an event scheduling system needs to process input dates, break them down into individual components (year, month, day), and return them along with the original formatted date string.


Problem Description
Implement the function process_date() that

Accepts a single argument date_str, a string representing a date date in the format "YYYY-MM-DD".

Returns a tuple containing the year, month, day, and the formatted date string.


Sample Input 1
"2023-02-04"


Sample Output 1
(2023, 2, 4, "2023-02-04")


Explanation
The sample input represents the date February 4, 2023. The output tuple contains the extracted year (2023), month (2), day (4), and the original formatted date string ("2023-02-04").

````python
from datetime import datetime

def process_date(date_str):

    dt = datetime.strptime(date_str, "%Y-%m-%d")

    return (
        dt.year,
        dt.month,
        dt.day,
        dt.strftime("%Y-%m-%d")
    )
````
