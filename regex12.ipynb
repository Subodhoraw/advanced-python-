## censor list  

```python
Hate_words = 'Awful','terrible','hate'
def censor_text(text):
    words = text.split()
    print(words)
    print(type(words))
    sensored = []
    for word in words:
        if word in Hate_words:
            sensored_word = '*' * len(word)
            sensored.append(sensored_word)
        else:
            sensored.append(word)
    censored_text = ' '.join(sensored)
    return censored_text
text = "i hate you"
print(censor_text(text))

```

Imagine you are developing a location management system for a company that operates multiple branches. Each branch has a unique location code composed of two uppercase letters followed by two digits, such as "AB12". The system needs to extract these location codes from various documents, emails, and reports to maintain an updated branch directory.


Problem Description
Implement the function extract_location_codes() that

Accepts a single argument text, a string containing various characters and sequences, with location codes mixed in.

Returns a list of strings, where each string is an extracted location code.


Notes
The location code strictly consists of two uppercase letters followed by two digits (e.g., "AB12").

Ignore any codes that do not match this format.


Sample Input 1

text = "Our new offices are located at GK12, AB34, and XY56. Visit us at GK78!"


```python
def extract_location_codes(text):
    #is easy to remember:

#[A-Z]{2} → exactly 2 uppercase letters
#\d{2} → exactly 2 digits
#\b ... \b → make sure it's a complete code, not part of a longer word.
    pattern = r"\b[A-Z]{2}\d{2}\b" 
    return re.findall(pattern,text)
```
