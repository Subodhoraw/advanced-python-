````python
Hate_words = 'Awful','terrible','hate'
def censor_text(text):
    words = text.split()
    print(words)
    print(type(words))
    sensored = []
    for word in words:
        if word in Hate_words:
            sensored_word = '*' * len(word)
            sensored.append(sensored_word)
        else:
            sensored.append(word)
    censored_text = ' '.join(sensored)
    return censored_text
text = "i hate you"
print(censor_text(text))

````

Accepts a single argument text, a string that represents text containing various words

Notes
Make use of regex patterns and use the re module.

"Hello there! My name is John and I love Python."


Sample Output 1
['Hello', 'My', 'John', 'I', 'Python']

Sample Output 2
['France', 'Germany']


Explanation
"NASA" and "USA" are ignored as they are fully uppercase, while "France" and "Germany" match the pattern.

````python
import re

def extract_capitalized_words(text):

    pattern = re.compile(r"\b(?:[A-Z][a-z]+|[A-Z])\b") # the update should be 

    return pattern.findall(text)
text = 'NASA and USA are abbreviations, but France and Germany are country names.'
print(extract_capitalized_words(text))

    
````

In today’s digital age, social media posts often contain hashtags that categorize content, making it easier for users to find related posts. Extracting these hashtags from a post is essential for analytics, trend tracking, and content management in various applications.


Problem Description
Implement a function extract_hashtags() that

Accepts a single argument post, a string containing text from a social media post, and

Returns a list of strings representing the valid hashtags found in the post.


Notes
A valid hashtag must start with # and can consist of letters (A-Z, a-z) and numbers (0-9).

Only standalone hashtags should be extracted (i.e., they must be separate words).

If no valid hashtags are found, return an empty list.


Sample Input 1
"Loving the #Python programming language! #Coding #100DaysOfCode"


Sample Output 1
["#Python", "#Coding", "#100DaysOfCode"]


Explanation
The post string contains three valid hashtags.


Sample Input 2
"No hashtags in this post. Just some text."


Sample Output 2
[]


Explanation
The post string has no valid hashtags, which results in an empty list.

````python
import re

def extract_hashtags(post):
    
    return re.findall(r'(?<!\S)#[A-Za-z0-9]+\b', post)
````
