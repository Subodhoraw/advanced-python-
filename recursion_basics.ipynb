```python
def sum_digi(n):
    if not n :
        return 0
    return (n % 10) + sum_digi(n // 10) # n % 10 = removes moddules, next n//10 it will return only the value    
print(sum_digi(10145))
```

def generate_permutations(s, current=""):
    # Base case
    if len(s) == 0:
        return [current]

    result = []

    # Try every character
    for i in range(len(s)):
        ch = s[i]

        # Remaining characters
        remaining = s[:i] + s[i+1:]

        # Recursive call
        result.extend(generate_permutations(remaining, current + ch))

    return result

```python
def count_char(s,c):
    if len(s) == 0:
       return 0
    if s[0]==c:
        return 1+ count_char(s[1:], c)
    else:
        return count_char(s[1:], c)

print(count_char("banana", "a"))
```
