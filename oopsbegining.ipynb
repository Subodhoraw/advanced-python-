```python
class TextProcessor:
    def to_lowercase(self,raw_text):
        return raw_text.lower()
    def remove_whitespace(self,raw_text):
        return raw_text.strip()

#1.instantiation: Building the physical objects 

processor = TextProcessor() ## object

print(id(processor))

#2. Utializing the Operator to access behaviour
dirty_text = "  Data science is awsome  "
print(dirty_text)
lower_text = processor.to_lowercase(dirty_text)
clean_text = processor.remove_whitespace(lower_text)

print(f"result: '{clean_text}'")
```
