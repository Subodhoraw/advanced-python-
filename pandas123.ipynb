## fixing 
Why is this happening?
The specific CSV file you are loading from that Gitlab link doesn't actually have a header row.

When you use pd.read_csv() without specifying the headers, pandas assumes the very first row of your actual data is the header. So, instead of having a column named 'Timestamp', your 0th column is likely named after the first date in the dataset (e.g., '2017-01-01 00:00:00').

The Fix
To solve this, we need to combine the logic from your earlier load_btc_market_price_with_names function with your current datetime parsing. You need to tell pandas that there is no header (header=None) and explicitly name the columns yourself using the names parameter.

Notes
If the DataFrame does not include the "Customer Type" column, the third returned DataFrame should be None.

The DataFrame should be modified only for sorting; the original DataFrame must remain unchanged.



````python
def sort_dataframe(df: pd.DataFrame):
    # 1. Sort by Purchase Amount (Highest to Lowest)
    df_by_amount = df.sort_values(by="Purchase Amount", ascending=False)
    print("--- Sorted by Purchase Amount (Highest to Lowest) ---")
    print(df_by_amount)
    print("\n")
    
    # 2. Sort by Customer Name (Alphabetical)
    df_by_name = df.sort_values(by="Customer Name", ascending=True)
    print("--- Sorted by Customer Name (Alphabetical) ---")
    print(df_by_name)
    print("\n")
    
    # 3. Sort by Customer Type ("Regular" before "VIP") then by Purchase Amount (Highest to Lowest)
    if "Customer Type" in df.columns:
        # Create a copy so we don't alter the original DataFrame's data types
        df_by_type_and_amount = df.copy()
        
        # Convert 'Customer Type' to a Categorical type to enforce the custom sorting order
        df_by_type_and_amount['Customer Type'] = pd.Categorical(
            df_by_type_and_amount['Customer Type'], 
            categories=["Regular", "VIP"], 
            ordered=True
        )
        
        # Sort using the categorical order (ascending) and then Purchase Amount (descending)
        df_by_type_and_amount = df_by_type_and_amount.sort_values(
            by=["Customer Type", "Purchase Amount"], 
            ascending=[True, False]
        )
        
        print("--- Sorted by Customer Type and Purchase Amount ---")
        print(df_by_type_and_amount)
        print("\n")
    else:
        df_by_type_and_amount = None
        print("--- Sorted by Customer Type and Purchase Amount ---")
        print("None (Column 'Customer Type' not found in DataFrame)")
        print("\n")
        
    return (df_by_amount, df_by_name, df_by_type_and_amount)
````

Use Case
In retail analytics, it is crucial to compute the total price for items purchased after including applicable taxes. A retail company has a dataset of transactions consisting of purchase amounts, and it wants to quickly compute the total price for each transaction after applying a tax rate. This computation should be performed on a DataFrame containing the data, making it easy to analyze and report on sales.


Problem Description
Implement a function calculate_total_price that:

Accepts a single parameter amount representing the purchase amount (can be an integer or a float).

Calculates the total price after applying tax based on the following rules:

If the amount is less than 100, apply a tax rate of 5%.

If the amount is between 100 and 200 (inclusive), apply a tax rate of 10%.

If the amount is greater than 200, apply a tax rate of 15%.

Returns the total price as a float.


Implement a function apply_total_price_tax that:

Accepts a single argument df, a pandas DataFrame containing a 'Purchase Amount' column.

Applies a tax rate on each purchase amount using apply() and calculate_total_price()

Prints and returns a new DataFrame with an added Total Price (After Tax) column which reflects these calculations.


Sample Input 1


````python
import pandas as pd

def calculate_total_price(amount):
    """
    Calculates the total price after applying the appropriate tax rate based on the amount.
    
    Args:
        amount (int or float): The initial purchase amount.
        
    Returns:
        float: The total price after adding the tax.
    """
    if amount < 100:
        # 5% tax
        total = amount + (amount * 0.05)
    elif 100 <= amount <= 200:
        # 10% tax
        total = amount + (amount * 0.10)
    else:
        # 15% tax
        total = amount + (amount * 0.15)
        
    return float(total)

def apply_total_price_tax(df):
    """
    Applies the calculate_total_price logic to the 'Purchase Amount' column.
    
    Args:
        df (pd.DataFrame): The input DataFrame containing a 'Purchase Amount' column.
        
    Returns:
        pd.DataFrame: The modified DataFrame with the new 'Total Price (After Tax)' column.
    """
    # Apply the custom function to each value in the 'Purchase Amount' column
    df['Total Price (After Tax)'] = df['Purchase Amount'].apply(calculate_total_price)
    
    # Print the modified DataFrame
    print(df)
    
    # Return the updated DataFrame
    return df
````

````python
def standardize_product_names(df):
    new_df = df.copy()
    
    # 2. Define the exact string replacements mapping dictionary
    replacement_rules = {
        'IPH 13': 'iPhone 13',
        'iPhone-13': 'iPhone 13',
        'Iphone 13': 'iPhone 13',
        'IPHONE_13': 'iPhone 13',
        'S21-Samsung': 'Samsung S21'
    }
    
    # 3. Apply the replacements to the 'Product Name' column
    new_df['Product Name'] = new_df['Product Name'].replace(replacement_rules)
    
    # 4. Print both DataFrames for verification
    print("--- Original DataFrame ---")
    print(df)
    
    print("\n--- Standardized DataFrame ---")
    print(new_df)
    
    # 5. Return the newly created DataFrame
    return new_df
````
