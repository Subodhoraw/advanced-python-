```python
import numpy as np
from scipy import stats

def analyze_prices(prices):
    prices = np.array(prices, dtype=float)

    # Remove NaN values
    prices = prices[~np.isnan(prices)]

    # If no valid values remain
    if prices.size == 0:
        return {
            "mean": None,
            "median": None,
            "mode": None
        }

    mean = np.mean(prices)
    median = np.median(prices)
    mode = stats.mode(prices, keepdims=True).mode[0]

    return {
        "mean": mean,
        "median": median,
        "mode": mode
    }
```

Concept Recap
In statistics, measures of central tendency (mean, median, and mode) summarize a dataset with a single representative value.

Mean: The sum of all values divided by the number of values.

Median: The middle value in a sorted list of numbers. If there is an even number of values, it’s the average of the two middle ones.

Mode: The most frequent value in the dataset.

Outliers (extreme values) can distort the mean, making it less representative. The median is more robust to outliers.


Problem Description
A real estate analyst needs to find the typical home price in a neighborhood, but the dataset includes unusually high prices that might skew the result.

Write a function find_central_tendency(prices) that:

Returns the mean if there are no outliers (prices ≤ $1,000,000).

Returns the median if there are outliers (any price > $1,000,000).

Returns the mode if all values are identical.

If all values are unique, return None for mode.

Outlier Detection:
Consider a price above $1,000,000 as an outlier.

If outliers are present, always return the median.

If no outliers, return the mean unless all values are identical.

```python
import numpy as np
from scipy import stats

def find_central_tendency(prices):
    prices = np.array(prices)

    if np.all(prices == prices[0]):
        result = stats.mode(prices, keepdims=True).mode[0]
        

    elif np.any(prices > 1_000_000):
        result = np.median(prices)

    else:
        result = np.mean(prices)

    return 
  
```

Problem Description
A movie recommendation system needs to estimate a missing movie rating for a user based on available ratings using basic matrix operations.

Implement a function predict_rating(ratings, user_index, movie_index) that:

Accepts a ratings matrix where each row represents a user and each column represents a movie's rating. Missing ratings are represented as np.nan.

Takes user_index, representing the target user.

Takes movie_index, representing the target movie.

Returns the predicted rating using the average rating of other users who have rated the movie.


Notes
If no other users have rated the movie, return np.nan.

Use NumPy functions for efficient computation.

```python
import numpy as np

def predict_rating(ratings, user_index, movie_index):
    ratings = np.array(ratings, dtype=float)

    # Get all ratings for the movie
    movie_ratings = ratings[:, movie_index]

    # Remove the target user's rating
    other_ratings = np.delete(movie_ratings, user_index)

    # Keep only valid (non-NaN) ratings
    valid_ratings = other_ratings[~np.isnan(other_ratings)]

    # If no valid ratings exist, return NaN
    if len(valid_ratings) == 0:
        return np.nan

    # Return the average rating
    return np.mean(valid_ratings)
```

```python
import numpy as np

def predict_rating_numpy(ratings, user_index, movie_index):
    ratings = np.array(ratings, dtype=float)

    # Get ratings for the specified movie
    movie_ratings = ratings[:, movie_index]

    # Exclude the target user's rating
    movie_ratings = np.delete(movie_ratings, user_index)

    # If all remaining ratings are NaN, return NaN
    if np.all(np.isnan(movie_ratings)):
        return np.nan

    # Compute the average ignoring NaN values
    return np.nanmean(movie_ratings)
```

Use Case
A sports coach wants to evaluate player performance based on multiple skill metrics to determine the best overall player.

Problem Description
Implement a function find_best_player(players) that:

Accepts a 2D NumPy array players, where each row represents a player and each column represents a skill metric.

Computes an overall performance score for each player.

Returns the index of the best-performing player.


Notes
Use NumPy operations such as dot product, arithmetic operations, and statistical functions where applicable.

The dataset may contain np.nan values, which should be ignored when computing scores.

In case of a tie, return the lowest index.

```python
import numpy as np

def find_best_player(players):
    players = np.array(players, dtype=float)

    # Compute the average score for each player, ignoring NaN values
    scores = np.nanmean(players, axis=1)

    # If a player's row is all NaN, nanmean returns NaN.
    # Treat those players as having the lowest possible score.
    scores = np.nan_to_num(scores, nan=-np.inf)

    # Return the index of the highest score.
    # np.argmax returns the first occurrence in case of a tie.
    return np.argmax(scores)
```
